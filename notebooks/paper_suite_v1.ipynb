```python
SOURCE_REF = 'main'
SOURCE_DIRECTORY = None  # 可选 Colab 内可读源码目录
PREVIOUS_RUNTIME_STOPPED = False  # 仅 public Baseline 的旧 runtime 已实际停止时改 True
```

若公开 Baseline 仍在其它 runtime 执行，本入口只报告 EXTERNAL_RUNTIME_ACTIVE 并继续独立部分，不抢占。只有实际停止旧任务后才设置上面的确认。正式结果全部在 Video-WM_Result；旧结果不删除、覆盖或重算。新同域对照与查询统计使用独立目录和预算。


```python
from pathlib import Path
import urllib.request
bootstrap_path = 'runtime/wan/paper_suite_colab_v1.py'
if SOURCE_DIRECTORY:
    bootstrap = (Path(SOURCE_DIRECTORY) / bootstrap_path).read_text()
else:
    bootstrap = urllib.request.urlopen(f'https://raw.githubusercontent.com/RICHAAARC/SC-SSTW/{SOURCE_REF}/{bootstrap_path}', timeout=120).read().decode('utf-8')
entry = {'__name__': 'paper_suite_colab_entry'}
exec(compile(bootstrap, bootstrap_path, 'exec'), entry)
RUN = entry['run_colab'](source_directory=SOURCE_DIRECTORY, source_ref=SOURCE_REF, previous_runtime_stopped=PREVIOUS_RUNTIME_STOPPED)
```
