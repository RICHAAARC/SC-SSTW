# Raw420 objective ablation: two fixed one-step arms

Click **Run all**. Both arms start from the same fixed POINT3/terminal.pt of run20261002T041230450908Z, after seven raw420 updates. This is the final point, not the earlier point with a better head margin. Previous update7 lowered the composite objective while worsening its global-worst term: improvement in the local mean exceeded the head-path deterioration. This motivates an objective ablation, not a claim that the local term or STE is already proven causal.

COMPOSITE retains global-worst hinge plus the mean of all173 pair hinges. WORST_ONLY retains only the global-worst hinge, still taking the minimum over all173 wrong classes with the same public template-derived margin. Both use eta174, shrink-only per-step L2cap1 and the original1392 pilot coefficients. No candidate-specific objective, local-weight scan, rank-rule change or payload change. Equal step rules do not imply equal actual step norms: both are retained and reported. Either result may be negative; neither arm is selected automatically.

A shared fresh native decode produces one baseline RGB8 raster. It is physically round-tripped through raw420 and raw444 and VAE-encoded as BASELINE_RAW420/BASELINE_RAW444. Both encoder VJPs use the exact same baseline raw420 received RGB, with only objective selection changed; both decoder VJPs use the same input terminal and native RGB. Encoder replay tensors/errors and decoder replay receipts are saved. Encoder and decoder causal checkpoint graphs run separately with the existing disk storage. Channel differentiation uses the same identity STE. COMPOSITE and WORST_ONLY are separate one-step branches, never sequential updates.

Each branch's full terminal is decoded, quantized and round-tripped through raw420 and raw444, then encoded and read. Six observations, two updates, twelve path reads, twelve payload reads,24 message comparisons and2088 valid costs. Three native decodes, six native encodes, two encoder/decoder VJP pairs, zero Transformer/MP4 calls. All174 candidates and all44 observations remain. Parent raw420/raw444 tensors and raw readouts, plus historical controls, are verified references, not new independent negatives. A failed arm stays in the denominator while the other arm can proceed.

Inspect actual minimum wrong-minus-true Delta, unique top/rank, raw444 contrast, repeated payload, objective components, actual update L2 and RGB change. Boundaries86/88 are reporting-only after complete blind readouts. The selected writer loss is recorded separately from the common composite loss diagnostic on every branch. Raw444 is a contrast, not a second writer objective. Baseline replay is reported without exact-environment or GPU-name requirements. Quality is versus the already-marked shared same-channel baseline, not OFF, and has no adopted acceptance threshold.

This fixed-source, fixed-phase terminal experiment does not establish generation-time trajectory embedding, unknown-phase synchronization, path-aligned payload closure, calibrated FPR or MP4 robustness. Local checks use CPU/static and simulated channel/model execution; full pretrained VAE and FFmpeg execution is left to the user. Select a GPU runtime and Run all. Uses current Python with fresh child processes and existing dependency repair. Results persist under MyDrive/Video-WM/Zero-Mean-Raw420-Objective-Ablation-V1/<timestamp>/fixed_reference/result.json.


```python
SOURCE_SHA = '07c90939da5823187f92ae7fa0dd8dc0b5488455'
from pathlib import Path
import datetime,json,os,subprocess,sys,traceback
STAMP=datetime.datetime.now(datetime.timezone.utc).strftime('%Y%m%dT%H%M%S%fZ')
OUTPUT=Path('/content/drive/MyDrive/Video-WM/Zero-Mean-Raw420-Objective-Ablation-V1')/STAMP
OUTPUT.mkdir(parents=True,exist_ok=False)
REPO=Path('/content/SC-SSTW-RAW420-OBJECTIVE-ABLATION-'+STAMP)
RUN_OUTPUT=OUTPUT/'fixed_reference'
PYTHON=sys.executable
INPUT_ROOT=Path('/content/drive/MyDrive/Video-WM/Zero-Mean-Raw420-Margin-Continue-V2/20261002T041230450908Z/fixed_reference')
REFERENCE_ROOT=Path('/content/drive/MyDrive/Video-WM/Zero-Mean-C1-YUV444-No-H264-V1/20261001T110124499275Z/fixed_reference')
setup=dict(status='SETUP_STARTED',source_sha=SOURCE_SHA,fixed_denominator={'updates': 2, 'new_observations': 6, 'path_reads': 12, 'payload_reads': 12, 'message_evaluations': 24, 'valid_costs': 2088, 'catalog_slots': 46980, 'structural_exclusions': 44892, 'reference_normalized': 3, 'reference_raw': 6, 'raw420_files': 3, 'raw444_files': 3, 'parent_normalized': 2, 'parent_raw': 4},observations={o:{'status':'NOT_RUN_SETUP'} for o in ['BASELINE_RAW420', 'BASELINE_RAW444', 'COMPOSITE_RAW420', 'COMPOSITE_RAW444', 'WORST_ONLY_RAW420', 'WORST_ONLY_RAW444']})
def write_json(path,value):
    temp=path.with_suffix(path.suffix+'.tmp');temp.write_text(json.dumps(value,indent=2)+'\n');os.replace(temp,path)
def failed(stage,exc):
    setup.update(status='SETUP_FAILED',stage=stage,error=f'{type(exc).__name__}: {exc}')
    write_json(OUTPUT/'setup_receipt.json',setup)
    write_json(OUTPUT/'setup_failure.json',dict(stage=stage,error=str(exc),traceback=traceback.format_exc()))
def logged(command,stage,cwd=None,check=True):
    try:
        with (OUTPUT/'execution.log').open('a') as log:
            log.write('COMMAND '+repr(command)+'\n');log.flush()
            child=subprocess.Popen(command,cwd=cwd,stdout=subprocess.PIPE,stderr=subprocess.STDOUT,text=True,bufsize=1)
            for line in child.stdout:print(line,end='',flush=True);log.write(line);log.flush()
            code=child.wait();log.write('EXIT '+str(code)+'\n')
        if check and code:raise subprocess.CalledProcessError(code,command)
        return code
    except Exception as exc:failed(stage,exc);raise
write_json(OUTPUT/'setup_receipt.json',setup)
print('Full writer terminal:',INPUT_ROOT,'Saved references:',REFERENCE_ROOT,'Output:',OUTPUT)
if SOURCE_SHA is None:
    error=RuntimeError('Unpublished draft: bind verified published source SHA before Run all')
    failed('SOURCE_BINDING',error);raise error
```
