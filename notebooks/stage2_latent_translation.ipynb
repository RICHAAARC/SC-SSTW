# Stage 2: common latent relocation — single GPU diagnostic

This notebook is a thin launcher for the included real module. CPU tests and static cell validation accompany the package; **GPU generation has not run**. Execute the GPU cell only after the separate one-run budget approval. This tests common spatial relocation, not temporal AISB writing, watermarks, or model trajectory control.

Fixed six arms: OFF1/OFF2/X+/X−/Y+/Y−. Wan2.1-1.3B, zero-based scheduler step 5, normal steps 6/7 remain; same .25 latent-pixel shift throughout time. Unmodified scheduler history is independently copied for each arm.


## 1. Fetch the fixed development branch
The repository is fetched directly; no source ZIP upload is required. The checked-out commit is printed before execution.

```python
from pathlib import Path
import subprocess
REPOSITORY_URL = 'https://github.com/RICHAAARC/SC-SSTW.git'
BRANCH = 'dev/真实视频公共观测/帧差加权质心-固定镜头单主体'
SOURCE = Path('/content/stage2_latent_translation_source')
if SOURCE.exists():
    raise FileExistsError('Source directory already exists; use a fresh runtime to avoid mixing versions')
subprocess.run(['git', 'clone', '--depth', '1', '--single-branch', '--branch', BRANCH, REPOSITORY_URL, str(SOURCE)], check=True)
subprocess.run(['git', '-C', str(SOURCE), 'rev-parse', 'HEAD'], check=True)
print(SOURCE)
```

## 2. Install the versioned lightweight runtime dependencies
Keep the Colab CUDA PyTorch installation. No model is loaded by this cell.

```python
import subprocess, sys
subprocess.run([sys.executable, '-m', 'pip', 'install',
    'diffusers==0.35.2', 'transformers', 'accelerate', 'ftfy',
    'sentencepiece', 'safetensors', 'huggingface_hub', 'numpy', 'Pillow'], check=True)
subprocess.run(['ffmpeg', '-version'], check=True, stdout=subprocess.DEVNULL)
```

## 3. Inspect the fixed plan
The run includes model loading in the 3600-second process-group limit. All fixed arms and failures remain in the output; no retry, amplitude/layer search, or model variants.

```python
import json
CONFIG = SOURCE / 'configs/stage2_latent_translation.json'
print(json.dumps(json.loads(CONFIG.read_text()), indent=2))
OUTPUT = Path('/content/drive/MyDrive/Video-WM/stage2-latent-translation/run01')
print('Output:', OUTPUT)
```

## 4. One approved GPU run
Set the flag only after the one-run GPU budget is approved. An existing output or execution log is never overwritten. The launcher enforces the wall limit, kills the worker process group on timeout, and saves the actual exit code.

```python
RUN_GPU_AFTER_APPROVAL = False
if not RUN_GPU_AFTER_APPROVAL:
    raise RuntimeError('GPU not executed: enable only after the one-run approval')
import os
env = dict(os.environ, PYTHONPATH=str(SOURCE), TOKENIZERS_PARALLELISM='false')
completed = subprocess.run([sys.executable, '-m', 'experiments.stage2.run_latent_translation',
    '--config', str(CONFIG), '--output', str(OUTPUT)], cwd=SOURCE, env=env)
exit_file = OUTPUT.parent / (OUTPUT.name + '.execution_exit.json')
print(exit_file.read_text() if exit_file.exists() else 'Launcher failed before an execution record was produced')
print('Launcher exit:', completed.returncode)
```
